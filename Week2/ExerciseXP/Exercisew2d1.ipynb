#Exercise 1: Introduction to Data Analysis
What is Data Analysis?
Data analysis is the systematic process of collecting, cleaning, transforming, and interpreting data to discover useful information, draw conclusions, and support decision-making. It involves applying statistical, logical, or computational techniques to understand patterns, trends, relationships, and outliers in datasets. The goal is to extract insights that can inform strategies, solve problems, or predict future outcomes.

Why Is Data Analysis Important in Modern Contexts?
In the modern world, data is generated at an unprecedented scale from sources such as social media, sensors, business transactions, and online activity. This abundance of data, often referred to as “big data,” can be meaningless without proper analysis.

Data analysis is crucial because it helps:

Make informed decisions: Companies and governments use data analysis to support policies, optimize operations, and improve services.

Identify problems early: Through real-time monitoring and trend detection, organizations can spot inefficiencies or risks before they escalate.

Gain competitive advantage: Businesses use data to understand customer behavior, forecast demand, and personalize experiences.

Support innovation: Insights from data fuel new products, services, and scientific discoveries.

Without data analysis, decision-making would be based on intuition or outdated information, which could lead to inefficiencies or missed opportunities.

Three Areas Where Data Analysis Is Applied Today
Healthcare

Data analysis in healthcare helps with diagnosing diseases, optimizing treatments, and predicting patient outcomes. For example, analyzing patient data can improve early detection of chronic conditions like diabetes or heart disease. Hospitals also use data analytics to manage resources and improve patient care quality.

Finance and Banking

In the financial sector, data analysis is used for fraud detection, credit scoring, and algorithmic trading. Banks analyze transaction data to flag suspicious activity, while credit card companies use it to assess risk before approving applications. Investment firms rely on predictive models to make data-driven trading decisions.

Marketing and E-Commerce

Businesses analyze customer data to create targeted marketing campaigns, recommend products, and optimize pricing strategies. For example, platforms like Amazon or Netflix use data analytics to suggest products or shows based on user behavior. This not only boosts sales but also enhances customer satisfaction.

Conclusion
Data analysis is a fundamental tool in today’s data-driven world. By turning raw data into actionable insights, it empowers individuals, organizations, and governments to make smarter decisions, improve outcomes, and stay competitive. As the amount of data continues to grow, the importance of data analysis across industries will only increase.

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Calculate statistics
mean_sepal_length = iris_df['SepalLengthCm'].mean()
median_sepal_length = iris_df['SepalLengthCm'].median()
mode_sepal_length = iris_df['SepalLengthCm'].mode()[0]  # mode() returns a Series

print(f"Mean: {mean_sepal_length}")
print(f"Median: {median_sepal_length}")
print(f"Mode: {mode_sepal_length}")

sns.boxplot(x='Species', y='SepalLengthCm', data=iris_df)
plt.title('Sepal Length Distribution by Species')
plt.xlabel('Species')
plt.ylabel('SepalLengthCm')
plt.show()
```

## Findings:

- The **mean** sepal length is approximately `5.84 cm`, the **median** is `5.8 cm`, and the **mode** is `5.0 cm`.
- The histogram shows a somewhat normal distribution, with most values between 5 and 6.5 cm.
- The boxplot indicates that *setosa* has shorter sepal lengths on average than *versicolor* and *virginica*.

These basic insights help understand the data’s central tendencies and variation.

```python
#Exercise 5 : Export a dataframe to excel format and JSON
# Step 1: Import necessary libraries
import pandas as pd

# Step 2: Create a simple DataFrame for students
data = {
    'Student_ID': [101, 102, 103, 104],
    'Name': ['Alice', 'Bob', 'Charlie', 'Diana'],
    'Grade': ['A', 'B', 'A', 'C'],
    'Age': [20, 21, 19, 22]
}

students_df = pd.DataFrame(data)

# Step 3: Export the DataFrame to an Excel file
excel_file = 'students_data.xlsx'
students_df.to_excel(excel_file, index=False)
print(f" Excel file exported successfully as '{excel_file}'")

# Step 4: Export the DataFrame to a JSON file
json_file = 'students_data.json'
students_df.to_json(json_file, orient='records', indent=4)
print(f" JSON file exported successfully as '{json_file}'")

# Step 5: Display the DataFrame
print("Students DataFrame Preview:")
print(students_df)
```

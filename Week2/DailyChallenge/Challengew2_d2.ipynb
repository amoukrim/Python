```python
#Daily Challenge: Hands-On Image Transformation and Visualization


from google.colab import files
#files.upload()  # Upload your kaggle.json here

#Import Libraries
import matplotlib.pyplot as plt
from PIL import Image, ImageOps
from scipy.ndimage import rotate
import numpy as np

# Load an image
image_path = './0095.png'
original_image = Image.open(image_path)

# Display the image
plt.imshow(original_image)
plt.title("Original Image")
plt.axis("off")
plt.show()
#Rotate the Image by 90 Degrees

def rotate_image_90_degrees(image):
    return rotate(image, angle=90, reshape=False)

# Convert to NumPy array and rotate
rotated_image = rotate_image_90_degrees(original_image)

# Show rotated image
plt.imshow(rotated_image)
plt.title("Rotated Image (90°)")
plt.axis("off")
plt.show()

 #Flip the Image Horizontally and Vertically

 # Flip horizontally
flipped_horizontally = ImageOps.mirror(original_image)

# Flip vertically
flipped_vertically = ImageOps.flip(original_image)

# Display both
fig, ax = plt.subplots(1, 2, figsize=(10, 4))
ax[0].imshow(flipped_horizontally)
ax[0].set_title("Flipped Horizontally")
ax[0].axis("off")

ax[1].imshow(flipped_vertically)
ax[1].set_title("Flipped Vertically")
ax[1].axis("off")

plt.show()

# Zoom by resizing
width, height = original_image.size
zoomed_image = original_image.resize((int(width * 1.2), int(height * 1.2)))

# Display zoomed image
plt.imshow(zoomed_image)
plt.title("Zoomed Image (1.2x)")
plt.axis("off")
plt.show()
```

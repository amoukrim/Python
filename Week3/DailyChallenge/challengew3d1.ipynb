#1 Préparation des Données

```python
#Création et préparation des données pour l'Analyse des températures dans 10 capitales européennes
#Préparation des Données
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Étape 1 : Création des données
# Liste de 10 capitales européennes
capitals = ["Paris", "Berlin", "Madrid", "Rome", "Vienna",
            "Amsterdam", "Brussels", "Lisbon", "Prague", "Copenhagen"]
            # Noms des mois pour les colonnes du DataFrame
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun",
          "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]

# Génération d'une graine aléatoire pour des résultats reproductibles
np.random.seed(42)

# Création d'un tableau de températures aléatoires entre -5°C et 35°C
# pour chaque ville (10 lignes) et chaque mois (12 colonnes)
temps = np.random.uniform(-5, 35, size=(10, 12))

# Création du DataFrame avec Pandas
df = pd.DataFrame(temps, index=capitals, columns=months)

# Étape 2 : Analyse
df["Moyenne Annuelle"] = df.mean(axis=1)

# Identification de la ville avec la température moyenne la plus élevée
ville_plus_chaude = df["Moyenne Annuelle"].idxmax()
# Identification de la ville avec la température moyenne la plus basse
ville_plus_froide = df["Moyenne Annuelle"].idxmin()

# Étape 3 : Visualisation
# Définition de la taille du graphique
plt.figure(figsize=(12, 7))

# Création d'une courbe de température pour chaque ville
for city in capitals:
    plt.plot(months, df.loc[city][months], label=city)
# Ajout de titres, légende et grille
plt.title("Températures moyennes mensuelles des capitales européennes")
plt.xlabel("Mois")
plt.ylabel("Température (°C)")
plt.grid(True)
plt.legend()
# Affichage du graphique
plt.show()

# Résumé
print("\n📈 Analyse du graphique :\n")

print(
    f"- La ville la plus chaude, {ville_plus_chaude}, présente des températures "
    f"élevées et relativement stables en été, avec des pics dépassant souvent les 30°C.\n"

    f"- En revanche, {ville_plus_froide} affiche des températures basses, notamment en hiver, "
    f"avec des minima proches de -5°C voire plus bas.\n"

    "- On observe que toutes les villes présentent une courbe en cloche typique du climat européen, "
    "avec des températures plus froides entre décembre et février, et un pic de chaleur en juillet/août.\n"

    "- Certaines capitales comme Lisbonne ou Rome montrent une montée rapide en température au printemps, "
    "tandis que des villes comme Copenhagen ou Berlin restent plus modérées.\n"

    "- Les différences de climat sont bien visibles : les capitales du sud de l'Europe ont des hivers doux "
    "et des étés très chauds, contrairement aux capitales du nord qui ont une amplitude thermique plus faible."
)
```

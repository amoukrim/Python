## 🌟 Exercice 2 : Statistiques descriptives
* Task: Given a sample dataset, calculate the mean, median, variance, and standard deviation using SciPy.
* Exemple de jeu de données :
data = [12, 15, 13, 12, 18, 20, 22, 21]


```python
# Import des fonctions statistiques depuis scipy
from scipy import stats
import numpy as np  # Utilisé ici pour la variance et l'écart type
```

```python
# Jeu de données fourni
data = [12, 15, 13, 12, 18, 20, 22, 21]
```

```python
# Moyenne : somme des valeurs divisée par le nombre de valeurs
mean = stats.tmean(data)
print("Moyenne :", mean)
```

```python
# Médiane : valeur centrale quand les données sont triées
median = stats.scoreatpercentile(data, 50)
print("Médiane :", median)
```

```python
# Variance : mesure de la dispersion des données autour de la moyenne
variance = np.var(data, ddof=0)  # ddof=0 → variance population
print("Variance :", variance)
```

```python
# Écart type : racine carrée de la variance
std_dev = np.std(data, ddof=0)  # ddof=0 → écart type population
print("Écart type :", std_dev)
```

## 🌟 Exercise 3: Understanding Distributions
* Task: Generate a normal distribution using SciPy with a mean of 50 and a standard deviation of 10. Plot this distribution..


```python
# Importation des bibliothèques nécessaires
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm
```

```python
# Paramètres de la distribution normale
mu = 50         # moyenne
sigma = 10      # écart type
```

```python
# Génération d'une plage de valeurs autour de la moyenne
x = np.linspace(mu - 4*sigma, mu + 4*sigma, 1000)
```

```python
# Calcul de la densité de probabilité pour chaque valeur de x
y = norm.pdf(x, mu, sigma)
```

```python
# Tracé de la courbe de la distribution normale
plt.plot(x, y)
plt.title("Distribution Normale (μ=50, σ=10)")
plt.xlabel("Valeurs")
plt.ylabel("Densité de probabilité")
plt.grid(True)
plt.show()
```

## 🌟 Exercise 4: T-Test Application
* Perform a T-test on two sets of randomly generated data.


```python
# Importation des bibliothèques nécessaires
import numpy as np
from scipy import stats
```

```python
# Génération de deux échantillons aléatoires suivant une distribution normale
# data1 : moyenne 50, écart type 10, 100 valeurs
# data2 : moyenne 60, écart type 10, 100 valeurs
data1 = np.random.normal(50, 10, 100)
data2 = np.random.normal(60, 10, 100)
```

```python
# Test t de Student pour échantillons indépendants
# H0 : les deux échantillons proviennent de populations avec la même moyenne
t_stat, p_value = stats.ttest_ind(data1, data2)
```

```python
# Affichage des résultats du test
print("Statistique t :", t_stat)
print("Valeur p :", p_value)
```


* Une p-value < 0.05 suggère un rejet de H0 (différence significative entre les moyennes)

## Exercise 5: Linear Regression Analysis
* Objective: Apply linear regression to a dataset and interpret the results.

* Task: Given a dataset of housing prices (house_prices) and their corresponding sizes (house_sizes), use linear regression to predict the price of a house given its size.
* Dataset:
house_sizes: [50, 70, 80, 100, 120] (in square meters)
house_prices: [150,000, 200,000, 210,000, 250,000, 280,000] (in currency units)
Questions:
What is the slope and intercept of the regression line?
Predict the price of a house that is 90 square meters.
Interpret the meaning of the slope in the context of housing prices.

```python
# Importation des bibliothèques nécessaires
import numpy as np
from scipy import stats
```

```python
# Données : tailles des maisons (m²) et prix correspondants
house_sizes = np.array([50, 70, 80, 100, 120])
house_prices = np.array([150000, 200000, 210000, 250000, 280000])
```

```python
# Régression linéaire : calcule la pente, l'interception, r, p-value, erreur std
slope, intercept, r_value, p_value, std_err = stats.linregress(house_sizes, house_prices)
```

```python
# Affichage des coefficients
print("Pente :", slope)
print("Interception :", intercept)
```

```python
# Prédiction du prix pour une maison de 90 m²
predicted_price = slope * 90 + intercept
print("Prix prédit pour une maison de 90 m² :", predicted_price)
```

### Interprétation :
* La pente représente l'augmentation moyenne du prix pour chaque mètre carré supplémentaire.
* Ici, chaque m² supplémentaire fait augmenter le prix d'environ `slope` unités monétaires.

## 🌟 Exercice 6 : Understanding ANOVA
* Objectif : Test understanding of ANOVA and its application.

Task: Three different fertilizers are applied to three separate groups of plants to test their effectiveness. The growth in centimeters is recorded.
Dataset:
fertilizer_1: [5, 6, 7, 6, 5]
fertilizer_2: [7, 8, 7, 9, 8]
fertilizer_3: [4, 5, 4, 3, 4]
Questions:
Perform an ANOVA test on the given data. What are the F-value and P-value?
Based on the P-value, do the fertilizers have significantly different effects on plant growth?
Explain what would happen if the P-value were greater than 0.05.

```python
# Importation de la fonction ANOVA (analyse de la variance) et numpy
from scipy import stats
import numpy as np
```

```python
# Données de croissance des plantes selon 3 types d'engrais
fertilizer_1 = [5, 6, 7, 6, 5]
fertilizer_2 = [7, 8, 7, 9, 8]
fertilizer_3 = [4, 5, 4, 3, 4]
```

```python
# Application du test ANOVA à un facteur (one-way ANOVA)
f_stat, p_value = stats.f_oneway(fertilizer_1, fertilizer_2, fertilizer_3)
```

```python
# Affichage des résultats
print("Valeur F :", f_stat)
print("Valeur p :", p_value)
```

### Interprétation :
* Si p < 0.05 : on rejette l'hypothèse nulle → au moins un engrais a un effet différent
* Si p > 0.05 : on ne peut pas conclure à une différence significative entre les engrais

```python
if p_value < 0.05:
    print("Résultat : Les engrais ont des effets significativement différents.")
else:
    print("Résultat : Aucune différence significative détectée entre les engrais.")
```

## Exercice 7 : Probability Distributions (Optional)
* Work with a binomial distribution: calculate probabilities for different numbers of successes.
Example: Calculating the probability of getting exactly 5 heads in 10 coin flips.

```python
# Importation de la distribution binomiale depuis scipy
from scipy.stats import binom
```

```python
# Paramètres de la distribution binomiale
n = 10      # nombre d'essais (lancers de pièce)
p = 0.5     # probabilité de succès (obtenir face)
```

```python
# Calcul de la probabilité d'obtenir exactement 5 faces
k = 5
prob = binom.pmf(k, n, p)  # pmf = probabilité de masse (discrète)

print(f"Probabilité d’obtenir exactement {k} faces sur {n} lancers :", prob)
```

## 🌟 Exercice 8 : Correlation Coefficients (Optional)
* Calculate the Pearson and Spearman correlation coefficients between two variables in a dataset.
Example in code: data = pd.DataFrame({'age': [23, 25, 30, 35, 40], 'income': [35000, 40000, 50000, 60000, 70000]})


```python
# Importation des bibliothèques nécessaires
import pandas as pd
from scipy.stats import pearsonr, spearmanr
```

```python
# Création de l'ensemble de données
data = pd.DataFrame({
    'age': [23, 25, 30, 35, 40],
    'income': [35000, 40000, 50000, 60000, 70000]
})
```

```python
# Calcul du coefficient de corrélation de Pearson (linéaire)
pearson_corr, _ = pearsonr(data['age'], data['income'])
print("Coefficient de corrélation de Pearson :", pearson_corr)
```

```python
# Calcul du coefficient de corrélation de Spearman (monotone)
spearman_corr, _ = spearmanr(data['age'], data['income'])
print("Coefficient de corrélation de Spearman :", spearman_corr)
```

Exercise 1 : Matrix Operations

```python

import numpy as np

# Step 1: Create a 3x3 matrix
matrix = np.array([
    [4, 7, 2],
    [3, 6, 1],
    [2, 5, 3]
])
print("Matrix:\n", matrix)

# Step 2: Calculate the determinant
determinant = np.linalg.det(matrix)
print("\nDeterminant:", round(determinant, 2))
# Step 3: Find the inverse (only if the determinant is not 0)
if determinant != 0:
    inverse = np.linalg.inv(matrix)
else:
    inverse = "Matrix is not invertible (determinant is 0)."

print("\nInverse:\n", inverse)
```

Exercise 4 : Data Manipulation with NumPy and Pandas

```python
import pandas as pd
import numpy as np

# Set random seed for reproducibility
np.random.seed(42)

#  Create DataFrame with 5 columns and 10 rows of random numbers
df = pd.DataFrame({
    'Integers': np.random.randint(0, 100, 10),
    'Decimals': np.random.uniform(0, 1, 10).round(2),
    'Negative_Nums': np.random.randint(-50, 50, 10),
    'Large_Nums': np.random.randint(1000, 10000, 10),
    'Percentages': np.random.uniform(0, 100, 10).round(2)
})

print("Original DataFrame:")
print(df)

# Conditional Selection Examples
print("\nConditional Selections:")

# Select rows where Integers > 50
print("\nRows where Integers > 50:")
print(df[df['Integers'] > 50])

# Select rows with negative values
print("\nRows with negative values:")
print(df[df['Negative_Nums'] < 0])

# Multiple conditions
print("\nRows where Integers > 30 AND Decimals < 0.5:")
print(df[(df['Integers'] > 30) & (df['Decimals'] < 0.5)])

# 3. Aggregation Functions
print("\nAggregation Results:")

# Basic aggregations
print("\nColumn sums:")
print(df.sum())

print("\nColumn averages:")
print(df.mean().round(2))

print("\nColumn maximums:")
print(df.max())

print("\ndescribe:", df.describe())


```

Exercise 5 : Image Representation

Images are represented as numerical arrays where:

Grayscale images are 2D arrays (height × width) with values typically ranging:

0 (black) to 255 (white) for 8-bit images

0.0 to 1.0 for floating-point representations

Color images are 3D arrays (height × width × channels), typically with:

3 channels (Red, Green, Blue) for RGB

4 channels (RGBA) when including alpha/transparency



```python
# voici un example pour créer une  image de  5×5 (une échelle de dégradé de gris)
# Manual Construction
import numpy as np
import matplotlib.pyplot as plt

# Create a 5×5 dégradé de gris
image = np.array([
    [0, 63, 127, 191, 255],    # Black to white gradient
    [31, 95, 159, 223, 255],   # Dark to light
    [127, 159, 191, 223, 255], # Medium to bright
    [63, 127, 191, 255, 255],  # Gradient with max white
    [0, 63, 127, 191, 255]     # Repeating gradient
], dtype=np.uint8)

plt.imshow(image, cmap='gray', vmin=0, vmax=255)
plt.colorbar()
plt.title("5×5 image avec une échelle de dégradé de gris")
plt.show()

```

Exercise 6 : Basic Hypothesis Testing

```python
#
import numpy as np

#Simulate the Data
# : Set a random seed for reproducibility so results are the same every time we run the code
np.random.seed(42)

# Step 2: Simulate employee productivity scores BEFORE the training
# Normally distributed with mean=50, standard deviation=10, for 30 employees
productivity_before = np.random.normal(loc=50, scale=10, size=30)

# Step 3: Simulate productivity scores AFTER the training
# Assume training improves productivity: we add a normal distribution centered around 5
productivity_after = productivity_before + np.random.normal(loc=5, scale=3, size=30)

# Step 4: Calculate average productivity before and after training
mean_before = np.mean(productivity_before)
mean_after = np.mean(productivity_after)

# Print the results
print(" Average Productivity Before Training:", round(mean_before, 2))
print(" Average Productivity After Training:", round(mean_after, 2))
print(" Average Increase in Productivity:", round(mean_after - mean_before, 2))

# Step 5: Compute the difference in productivity for each employee (after - before)
productivity_diff = productivity_after - productivity_before

# Step 6: Calculate the average improvement and standard deviation of the improvement
# This helps us understand not just the average change, but also how consistent the change is
mean_diff = np.mean(productivity_diff)
std_diff = np.std(productivity_diff)

# Print the improvement statistics
print("\n Mean Improvement per Employee:", round(mean_diff, 2))
print(" Standard Deviation of Improvement:", round(std_diff, 2))
```

conclusion de l'exercice 6 : D'après ces données simulées, la productivité moyenne(the average productivity) a augmenté d'environ 5 points, et l'écart type (and the standard deviation) suggère que l'amélioration est relativement constante parmi les employés. Cela témoigne de l'efficacité du programme de formation.

Exercise 8 : Time Series Data Manipulation

```python
import numpy as np
import pandas as pd

# Période : Tous les jours de 2023
dates = pd.date_range(start="2023-01-01", end="2023-12-31", freq="D")

# Données aléatoires (ex: température quotidienne)
np.random.seed(42)
data = np.random.randint(low=0, high=35, size=len(dates))  # Températures entre 0°C et 35°C

# Création d'un DataFrame Pandas
ts = pd.DataFrame({"Date": dates, "Temperature": data})
ts.set_index("Date", inplace=True)

print(ts.head())
# utilisation du slicing  pour le découpage par trimestres
#de Janvier à Mars

jan_mar = ts["2023-01":"2023-03"]
print("\n de Janvier à Mars:\n", jan_mar.head())


# de April à juin
apr_jun = ts["2023-04":"2023-06"]
print("\nAvril à Juin (Q2):\n", apr_jun.head())
# de Juillet à SEptembre
jul_sep = ts["2023-07":"2023-09"]
print("\nJuillet à Septembre (Q3):\n", jul_sep.head())

#de Octobre à décembre
oct_dec = ts["2023-10":"2023-12"]
print("\nOctobre à Décembre (Q4):\n", oct_dec.head())
```

```python
# convert a NumPy array to a Pandas DataFrame

import numpy as np
import pandas as pd

# Créer un array NumPy
data = np.array([
    [1, "Alice", 25],
    [2, "Bob", 30],
    [3, "Charlie", 35]
])

# Convertir en DataFrame
df = pd.DataFrame(data, columns=["ID", "Name", "Age"])
print("DataFrame:\n", df)

#Conversion d'un Pandas DataFrame en NumPy Array

# Reprendre le DataFrame précédent
array_from_df = df.to_numpy()
print("\nNumPy Array:\n", array_from_df)
```

Exercise 10 : Basic Visualization

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.ticker import MaxNLocator

# Set random seed for reproducibility
np.random.seed(42)

#  Create DataFrame with 5 columns and 10 rows of random numbers

mois = np.array(["janvier", "Février", "Mars","Avril", "Mai", "Juin", "Juillet", "Août", "Septembre", "Octobre"])
valeurs = np.random.randint(0, 100, size=len(mois))

# Création de la figure avec une taille optimisée
fig, ax = plt.subplots(figsize=(12, 6), dpi=100)

# Tracé avec améliorations
line = ax.plot(mois, valeurs,
               marker='o',
               markersize=8,
               linewidth=2.5,
               color='royalblue',
               markerfacecolor='white',
               markeredgewidth=2)

# Personnalisation des axes et du titre
ax.set_xlabel('Mois', fontsize=12, labelpad=10, fontweight='bold')
ax.set_ylabel('Consommation (MV)', fontsize=12, labelpad=10, fontweight='bold')
ax.set_title('Consommation électrique mensuelle (2023)',
             fontsize=14, pad=20, fontweight='bold')

# Amélioration des ticks et grille
ax.xaxis.set_tick_params(labelsize=10, rotation=45)
ax.yaxis.set_major_locator(MaxNLocator(integer=True))
ax.grid(True, linestyle='--', alpha=0.7)

# Ajout des valeurs sur les points
for i, v in enumerate(valeurs):
    ax.text(i, v+3, str(v),
            ha='center',
            va='bottom',
            fontsize=8,
            fontweight='bold')

# Ajustement des limites et espacement
ax.set_ylim(0, max(valeurs)*1.15)
plt.tight_layout()


ax.grid()

plt.show()
```
